```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

df=pd.read_csv("StudentsPerformance.csv")
df
```

#change math score from int to float

```python
df["math score"]=df["math score"].replace(" ","0")
df["math score"]=df["math score"].astype("float")
```

```python
ax=sns.countplot(x="math score",data=df)
ax.bar_label(ax.containers[0])
plt.title('score count')
plt.show()
sns.countplot(x="reading score",data=df)
plt.title('score count')
plt.show
sns.countplot(x="writing score",data=df)
plt.show
```

```python
plt.figure(figsize=(10,1000))
gb=df.groupby('math score').agg({'math score':'count'})

plt.pie(gb['math score'],labels =gb.index)
plt.title('score count')
plt.show()

# plt.pie(df['reading score'])
# plt.show()
# plt.pie(df['writing score'])
# plt.show()
```

```python
sns.countplot(x='gender',data=df, hue='math score')
plt.title('Score By Gender')
plt.show()
```

```python
sns.countplot(x='math score',data=df, hue='')
plt.title('Score By math')
plt.show()
```

```python
sns.boxplot(x="gender", y="math score", data=df)
plt.title("Gender vs Math Score")
plt.show()
```

```python
sns.boxplot(x="test preparation course", y="math score", data=df)
plt.title("test preparation course vs Math Score")
plt.show()
```

```python
sns.boxplot(x="parental level of education", y="race/ethnicity", data=df)
plt.title("parental level of education vs race/ethnicity")
plt.show()
```

```python
numeric_df = df.select_dtypes(include='number')
numeric_df.head()
```

```python
corr_matrix = numeric_df.corr()
corr_matrix
```

```python
sns.heatmap(corr_matrix, annot=True)
plt.title("Correlation Matrix")
plt.show()
```

```python
sns.pairplot(df, hue="gender")
plt.show()
```

```python
df.columns.values
```

```python
plt.figure(figsize=(15, 10))
cat_cols = [
    'gender',
    'race/ethnicity',
    'parental level of education',
    'lunch',
    'test preparation course'
]
for i, col in enumerate(cat_cols, 1):
    plt.subplot(2, 3, i)
    sns.countplot(x=col, data=df, hue=df['gender'])
    plt.title(col)
    plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
sns.scatterplot(
    x="math score",
    y="reading score",
    hue="gender",
    data=df)

plt.title("Math Score vs Reading Score by Gender")
plt.show()
```

```python
sns.scatterplot(
    x="math score",
    y="writing score",
    hue="gender",
    data=df
)

plt.title("Math Score vs writing score by Gender")
plt.show()
```

```python
sns.scatterplot(
    x="writing score",
    y="reading score",
    hue="gender",
    data=df
)

plt.title("writing Score vs Reading Score by Gender")
plt.show()
```

```python
sns.pairplot(
    df,
    vars=["math score", "reading score", "writing score"],
    hue="gender")
plt.show()
```

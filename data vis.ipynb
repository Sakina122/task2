```python


# Load data
df = pd.read_csv("StudentsPerformance.csv")

# Create total score
df["total_score"] = df["math score"] + df["reading score"] + df["writing score"]

# Select score columns
score_cols = ["math score", "reading score", "writing score", "total_score"]

# Correlation matrix
corr_matrix = df[score_cols].corr()
print("Correlation Matrix:\n", corr_matrix)

# Heatmap
plt.imshow(corr_matrix)
plt.colorbar()
plt.xticks(range(len(score_cols)), score_cols, rotation=45)
plt.yticks(range(len(score_cols)), score_cols)
plt.title("Correlation Heatmap (Score Columns)")
plt.show()

# Strongest positive correlation
corr_no_diag = corr_matrix.where(~np.eye(corr_matrix.shape[0], dtype=bool))
pair = corr_no_diag.stack().idxmax()
value = corr_no_diag.stack().max()

print("Strongest Positive Correlation:", pair)
print("Correlation Value:", value)

```
